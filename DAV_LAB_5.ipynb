```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
df = pd.read_csv("melb_data.csv")
```

```python
print("Original Dataset:")
print(df.head())
```

```python
print("\nShape of dataset:")
print(df.shape)
```

```python
print("\nColumn names:")
print(df.columns)
```

```python
print("\nDataset information:")
print(df.info())
```

```python
print("\nStatistical summary:")
print(df.describe())
```

```python
print("\nMisssing Values Before Preprocessing:")
print(df.isnull().sum())
```

```python
df_mean = df.copy()
```

```python
mean_building_area = df_mean['BuildingArea'].mean()
```

```python
df_mean['BuildingArea'] = df_mean['BuildingArea'].fillna(mean_building_area)
```

```python
print("\nMisssing Values After Preprocessing:")
print(df_mean.isnull().sum())
```

```python
print("\nMean building Area:",mean_building_area)
```

```python
df_median = df.copy()
```

```python
median_year_built = df_median['YearBuilt'].median()
```

```python
df_median["YearBuilt"] = df_median["YearBuilt"].fillna(median_year_built)
```

```python
print("\nMeadian Year Built:",median_year_built)
```

```python
print("Missing YearBuilt Values After median imputation:")
print(df_median["YearBuilt"].isnull().sum())
```

```python
df_ffill = df.copy()
df_ffill["Car"] = df_ffill["Car"].fillna(method="ffill")
print("\nMissing Car Values After forward fill:")
print(df_ffill["Car"].isnull().sum())
```

```python
df_processed = df.copy()

council_mode = df_processed["CouncilArea"].mode()[0]

df_processed["CouncilArea"] = df_processed["CouncilArea"].fillna(
    council_mode
)

print("\nCouncilArea Mode: ", council_mode)

print("\nMissing Categorical Values After Imputation:")
print(df_processed["CouncilArea"].isnull().sum())
```

```python
df_encoded = pd.get_dummies(
    df_processed,
    columns=["Type", "Method"],
    dtype=int
)

print("\nDataset After One-Hot Encoding:")
print(df_encoded.head())


# Label Encoding
df_label = df.copy()

label_encoder = LabelEncoder()

df_label["CouncilArea_Encoded"] = label_encoder.fit_transform(
    df_label["CouncilArea"]
)

print("\nLabel Encoding of CouncilArea:")
print(
    df_label[
        ["CouncilArea", "CouncilArea_Encoded"]
    ].head(10)
    )
```
